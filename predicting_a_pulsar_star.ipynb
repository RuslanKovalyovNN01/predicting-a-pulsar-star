```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
data = pd.read_csv("sample_data/pulsar_stars.csv")
```

```python
header = ['mean_pr', 'std_pr', 'kurtosis_pr', 'skewness_pr',
           'mean_dmsnr_c', 'std_dmsnr_c', 'kurtosis_dmsnr_c',
           'skewness_dmsnr_c', 'class']
data.columns = header
data
```

## Построение матрицы ковариации

```python
from pandas.plotting import scatter_matrix
scatter_matrix(data, alpha = .01 , figsize = (20,20))
plt.tight_layout()
pass
```

```python
import seaborn as sns
f, ax = plt.subplots(figsize=(15,15))
sns.heatmap(data.corr(), annot = True, linecolor = "blue", fmt = ".2", ax = ax)
plt.show()
```

```python
plt.plot(data['kurtosis_pr'], data['skewness_pr'], 'o', alpha = 0.05)
plt.xlabel('kurtosis_profile')
plt.ylabel('skewness_profile')
plt.title('Excess kurtosis of the integrated profile vs Skewness of the integrated profile')
pass
```
